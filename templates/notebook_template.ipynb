## 3 – Clean & Transform

```python
df = df_raw.copy()

# Normalise column names
df = normalize_columns(df)

# Strip string whitespace
df = strip_strings(df)

# Remove duplicates and fully-null columns
df = drop_duplicates_and_nulls(df)

# Fill missing numeric values with column median
df = fill_missing(df, strategy='median')

# TODO: add assignment-specific transformations here
# df = parse_dates(df, ['date_column'])
# df = filter_rows(df, 'score', '>=', 80)
# df = add_derived_column(df, 'score_pct', lambda d: d['score'] / 100)

df.head()
```

## 4 – Validate

```python
assert not df.empty, 'DataFrame must not be empty'

# TODO: add required columns for your assignment
required_cols = ['id', 'name', 'score']
missing = [c for c in required_cols if c not in df.columns]
assert not missing, f'Missing columns: {missing}'

null_pct = df.isnull().mean() * 100
print('Null % per column:')
print(null_pct[null_pct > 0].to_string() or '  (none)')

print('\nValidation passed ✓')
```
